```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# Configuração da figura e do plano de fundo
fig, ax = plt.subplots(figsize=(10, 8), dpi=300)
ax.set_facecolor('#f8f9fa')
fig.patch.set_facecolor('#f8f9fa')

# Título do Diagrama
plt.text(5, 9.5, "Fluxo de Trabalho - Vigilância Sanitária Municipal",
         ha='center', va='center', fontsize=14, fontweight='bold', color='#0f766e')

# Função auxiliar para desenhar caixas de texto estilo cartão
def draw_box(x, y, w, h, text_title, text_sub="", bg_color="#0d9488"):
    rect = patches.FancyBboxPatch((x - w/2, y - h/2), w, h,
                                 boxstyle="round,pad=0.02,rounding_size=0.15",
                                 linewidth=0, facecolor=bg_color)
    ax.add_patch(rect)
    if text_sub:
        ax.text(x, y + 0.12, text_title, color='white', weight='bold', fontsize=10, ha='center', va='center')
        ax.text(x, y - 0.15, text_sub, color='white', fontsize=8, ha='center', va='center')
    else:
        ax.text(x, y, text_title, color='white', weight='bold', fontsize=10, ha='center', va='center')

# Desenhando as Caixas do Fluxograma
draw_box(5, 8.5, 6, 0.7, "Solicitação de Licença / Denúncia / Rotina", bg_color="#0d9488")
draw_box(5, 7.0, 6.5, 0.7, "Análise Documental & Agendamento de Inspeção", bg_color="#0f766e")
draw_box(5, 5.3, 7, 0.9, "Inspeção Sanitária In Loco (Avaliação de Risco)", "(Estrutura, PGRSS, Processos e Equipamentos)", bg_color="#115e59")

# Caixas do Split (Conformidade vs Inconformidade)
draw_box(2.8, 3.4, 4, 0.9, "Concessão da Licença Sanitária", "(Estabelecimento em Conformidade)", bg_color="#16a34a")
draw_box(7.2, 3.4, 4, 0.9, "Auto de Infração / Notificação", "(Prazo para Adequação ou Interdição)", bg_color="#dc2626")

# Caixa Final
draw_box(5, 1.5, 6.5, 0.9, "Ações Educativas e Mitigação de Riscos", "(Garantia da Segurança do Paciente e do Trabalhador)", bg_color="#1d4ed8")

# Desenhando as Setas de Conexão
arrow_props = dict(arrowstyle="->", color="#475569", lw=1.5)

# Setas do fluxo principal
ax.annotate("", xy=(5, 7.35), xytext=(5, 8.15), arrowprops=arrow_props)
ax.annotate("", xy=(5, 5.75), xytext=(5, 6.65), arrowprops=arrow_props)

# Bifurcação (Com caminhos em L)
ax.plot([5, 2.8, 2.8], [4.85, 4.85, 3.85], color="#475569", lw=1.5)
ax.annotate("", xy=(2.8, 3.85), xytext=(2.8, 3.9), arrowprops=arrow_props)

ax.plot([5, 7.2, 7.2], [4.85, 4.85, 3.85], color="#475569", lw=1.5)
ax.annotate("", xy=(7.2, 3.85), xytext=(7.2, 3.9), arrowprops=arrow_props)

# Junção para a etapa final
ax.plot([2.8, 2.8, 5], [2.95, 2.3, 2.3], color="#475569", lw=1.5)
ax.plot([7.2, 7.2, 5], [2.95, 2.3, 2.3], color="#475569", lw=1.5)
ax.annotate("", xy=(5, 1.95), xytext=(5, 2.3), arrowprops=arrow_props)

# Limites da tela
ax.set_xlim(0, 10)
ax.set_ylim(0, 10)
ax.axis('off')

# Salvar como imagem PNG de alta qualidade
plt.tight_layout()
plt.savefig("fluxograma_vigilancia_sanitaria.png", dpi=300, bbox_inches='tight')
print("Imagem 'fluxograma_vigilancia_sanitaria.png' gerada com sucesso!")
```
